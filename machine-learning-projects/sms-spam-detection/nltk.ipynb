```python
import requests

# Получение стоп-слов с удаленного ресурса
stopwords_url = "https://raw.githubusercontent.com/nltk/nltk_data/gh-pages/packages/corpora/stopwords.zip"
response = requests.get(stopwords_url)

# Сохранение содержимого в локальный файл
with open("stopwords.zip", "wb") as file:
    file.write(response.content)

# Распаковка ZIP-архива
import zipfile
with zipfile.ZipFile("stopwords.zip", "r") as zip_ref:
    zip_ref.extractall("./nltk_data/corpora")

# Использование стоп-слов из NLTK
from nltk.corpus import stopwords
print(stopwords.words('russian'))
```
